### Which drug has sold the most often in 2017? ###

### Which drug category has the highest average daily sales? ###

Use mean() to calculate the average daily sales of each drug category.

```python
# Calculate average daily sales
avg_sales = df[drug_cols].mean()
# Sort in descending order
print(avg_sales.sort_values(ascending=False))
```

```python
# Create bar chart (daily average sales)
plt.figure(figsize=(9, 5))
plt.bar(avg_sales.index, avg_sales.values)

plt.title("Average Daily Sales of Drug Categories")
plt.xlabel("Drug Category")
plt.ylabel("Average Daily Sales")

plt.xticks(rotation=45)
plt.show()
```
